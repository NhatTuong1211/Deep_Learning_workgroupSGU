```python
import data
import plot_data
 
 
def smoke_test_dataloader():
    try:
        import torch
    except ImportError:
        print("[BO QUA] Chua cai torch -> chua kiem tra duoc dataset.py")
        return
    from dataset import get_dataloader
 
    x, y = next(iter(get_dataloader("train", batch_size=32)))
    assert tuple(x.shape) == (32, 2) and tuple(y.shape) == (32, 1), "sai shape batch"
    assert x.dtype == torch.float32 and y.dtype == torch.float32, "sai dtype"
 
    sizes = {"train A=200": get_dataloader("train", 32, train_size=200).dataset,
             "val": get_dataloader("val", 32).dataset,
             "test": get_dataloader("test", 32).dataset}
    assert [len(d) for d in sizes.values()] == [200, 200, 100], "sai so mau"
 
    a = next(iter(get_dataloader("train", 32, seed=7)))[0]
    b = next(iter(get_dataloader("train", 32, seed=7)))[0]
    assert torch.equal(a, b), "cung seed nhung khac thu tu batch"
    print("[OK] DataLoader: shape (32,2)/(32,1), float32, dung kich thuoc tap, tai lap theo seed.")
 
 
def main():
    print("== Buoc 1: du lieu ==")
    data.main()
    print("\n== Buoc 2: hinh ve ==")
    plot_data.main()
    print("\n== Buoc 3: DataLoader ==")
    smoke_test_dataloader()
 
 
if __name__ == "__main__":
    main()
```
